```python
dfs = []

for j in range(1, 20):
    url = 'https://www.ambitionbox.com/list-of-companies?page={}'.format(j)
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage = requests.get(url, headers=headers).text
    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('div', class_='companyCardWrapper')
    
    name = []
    rating = []
    role = []
    
    for i in company:
        comp_name = i.find('h2', class_='companyCardWrapper__companyName')
        comp_rating = i.find('div', class_='rating_star_container')
        comp_role = i.find('div', class_='companyCardWrapper__ratingHeader')
        
        name.append(comp_name.text.strip() if comp_name else None)
        rating.append(comp_rating.text.strip() if comp_rating else None)
        role.append(comp_role.text.strip() if comp_role else None)
    
    d = {'name': name, 'rating': rating, 'role': role}
    dfs.append(pd.DataFrame(d))

final = pd.concat(dfs, ignore_index=True)
```
