```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('tested_titanic.csv')
df = df[['Age', 'Pclass', 'SibSp', 'Parch', 'Fare', 'Survived']]
```

```python
df.head()
```

```python
print(df.isnull().sum())
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2
)
```

```python
knn = KNNImputer(n_neighbors=3, weights='distance')
```

```python
X_train_trf = knn.fit_transform(X_train)
```

```python
X_test_trf = knn.transform(X_test)
```

```python
X_train_trf_df = pd.DataFrame(X_train_trf, columns=X_train.columns)
print(X_train_trf_df.isnull().sum())
```

```python
lr = LogisticRegression()
lr.fit(X_train_trf, y_train)
y_pred = lr.predict(X_test_trf)
accuracy_score(y_test, y_pred)
```

```python
clf = LogisticRegression()
clf.fit(X_train_trf, y_train)

y_pred = clf.predict(X_test_trf)
print("Accuracy with KNN Imputer:", accuracy_score(y_test, y_pred))
```

```python
si = SimpleImputer(strategy='mean')

X_train_si = si.fit_transform(X_train)
X_test_si = si.transform(X_test)

clf_si = LogisticRegression()
clf_si.fit(X_train_si, y_train)

y_pred_si = clf_si.predict(X_test_si)
print("Accuracy with Simple Imputer (Mean):", accuracy_score(y_test, y_pred_si))
```
