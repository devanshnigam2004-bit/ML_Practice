```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('tested_titanic.csv')
df = df[['Age', 'Fare', 'Survived']]
```

```python
df.sample(5)
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2
)
```

```python
# Create an imputed column with original values
X_train['Age_imputed'] = X_train['Age']
X_test['Age_imputed'] = X_test['Age']
```

```python
# Extract random samples from non-null values to fill missing values
num_missing_train = X_train['Age_imputed'].isnull().sum()
random_sample_train = (
    X_train['Age'].dropna().sample(num_missing_train, random_state=0)
)
random_sample_train.index = X_train[X_train['Age_imputed'].isnull()].index
X_train.loc[X_train['Age_imputed'].isnull(), 'Age_imputed'] = (
    random_sample_train
)

num_missing_test = X_test['Age_imputed'].isnull().sum()
random_sample_test = (
    X_train['Age']
    .dropna()
    .sample(num_missing_test, random_state=0)
)  # Note: sample from train set
random_sample_test.index = X_test[X_test['Age_imputed'].isnull()].index
X_test.loc[X_test['Age_imputed'].isnull(), 'Age_imputed'] = random_sample_test
```

```python
data = pd.read_csv('train.csv', usecols=['GarageQual', 'FireplaceQu', 'SalePrice'])
```

```python
data.head()
```

```python
data.isnull().mean() * 100
```

```python
X = data
y = data['SalePrice']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2
)
```

```python
X_train['GarageQual_imputed'] = X_train['GarageQual']
X_test['GarageQual_imputed'] = X_test['GarageQual']

X_train['FireplaceQu_imputed'] = X_train['FireplaceQu']
X_test['FireplaceQu_imputed'] = X_test['FireplaceQu']
```

```python
X_train.sample(5)
```

```python
X_train['GarageQual_imputed'][
    X_train['GarageQual_imputed'].isnull()
] = X_train['GarageQual'].dropna().sample(
    X_train['GarageQual'].isnull().sum()
).values
X_test['GarageQual_imputed'][
    X_test['GarageQual_imputed'].isnull()
] = X_train['GarageQual'].dropna().sample(
    X_test['GarageQual'].isnull().sum()
).values

X_train['FireplaceQu_imputed'][
    X_train['FireplaceQu_imputed'].isnull()
] = X_train['FireplaceQu'].dropna().sample(
    X_train['FireplaceQu'].isnull().sum()
).values
X_test['FireplaceQu_imputed'][
    X_test['FireplaceQu_imputed'].isnull()
] = X_train['FireplaceQu'].dropna().sample(
    X_test['FireplaceQu'].isnull().sum()
).values
```

```python
# GarageQual_imputed ke liye (Train set)
train_missing_mask = X_train['GarageQual_imputed'].isnull()
train_samples = (
    X_train['GarageQual']
    .dropna()
    .sample(train_missing_mask.sum(), random_state=0)
)
X_train.loc[train_missing_mask, 'GarageQual_imputed'] = train_samples.values

# GarageQual_imputed ke liye (Test set)
test_missing_mask = X_test['GarageQual_imputed'].isnull()
test_samples = (
    X_train['GarageQual']
    .dropna()
    .sample(test_missing_mask.sum(), random_state=0)
)
X_test.loc[test_missing_mask, 'GarageQual_imputed'] = test_samples.values

# FireplaceQu_imputed ke liye (Train set)
train_fire_mask = X_train['FireplaceQu_imputed'].isnull()
train_fire_samples = (
    X_train['FireplaceQu'].dropna().sample(train_fire_mask.sum(), random_state=0)
)
X_train.loc[train_fire_mask, 'FireplaceQu_imputed'] = train_fire_samples.values

# FireplaceQu_imputed ke liye (Test set)
test_fire_mask = X_test['FireplaceQu_imputed'].isnull()
test_fire_samples = (
    X_train['FireplaceQu'].dropna().sample(test_fire_mask.sum(), random_state=0)
)
X_test.loc[test_fire_mask, 'FireplaceQu_imputed'] = test_fire_samples.values
```
